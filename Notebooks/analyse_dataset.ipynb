Objectif : comprendre les données
- Vérifier la taille des images
- détecter les doublons
- voir quelques exemples
- analyser la distribution

```python
import os
import random
from PIL import Image
import matplotlib.pyplot as plt

DATASET_PATH = "../datasets/cleaned"

files = [f for f in os.listdir(DATASET_PATH) if f.lower().endswith(".jpg")]

print("Nombre total d'images :", len(files))

sizes = {}

for file in files:
    
    img_path = os.path.join(DATASET_PATH, file)

    try:
        with Image.open(img_path) as img:
            size = img.size
            sizes[size] = sizes.get(size, 0) + 1
    except:
        continue

print("Distribution des tailles d'images :")
print(sizes)
```

### Afficher quelques images

```python
sample = random.sample(files, 9)

plt.figure(figsize=(8,8))

for i, file in enumerate(sample):
    
    img_path = os.path.join(DATASET_PATH, file)
    img = Image.open(img_path)

    plt.subplot(3,3,i+1)
    plt.imshow(img)
    plt.axis("off")

plt.show()
```

### Vérifier la taille idéale (224×224)

```python
wrong_size = []

for file in files:

    img_path = os.path.join(DATASET_PATH, file)

    try:
        with Image.open(img_path) as img:
            if img.size != (224,224):
                wrong_size.append(file)
    except:
        continue

print("Nombre d'images non 224x224 :", len(wrong_size))
```
